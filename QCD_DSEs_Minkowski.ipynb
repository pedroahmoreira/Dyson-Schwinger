## Solução das Equações de Dyson-Schwinger no Espaço de Minkowski

### Propagador do Férmion
O propagador do férmion vestido é dado por:
\begin{equation}
    S_f(k) = \frac{1}{k\!\!\!/ - m_B + k\!\!\!/A_f(k^2) - B_f(k^2) + i\varepsilon}
\end{equation}
ou
\begin{equation}
    S_f^{-1}(k) = k\!\!\!/ - m_B + k\!\!\!/A_f(k^2) - B_f(k^2) + i\varepsilon
\end{equation}
e a auto-energia pela representação integral
