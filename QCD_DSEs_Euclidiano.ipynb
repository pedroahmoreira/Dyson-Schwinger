## Equações de Dyson-Schwinger da Cromodinâmica Quântica no espaço euclidiano

A teoria em foco é a Cromodinâmica quântica (QCD), uma teoria de calibre com simetria $SU(3)$ que descreve as interações fortes. Sua densidade lagrangiana completa, no espaço de coordendadas, é
\begin{equation}
    \mathcal{L}_{QCD} = \bar{q}_{ij} (i \gamma^{\mu} (D_{\mu})_{ij} - m \delta_{ij}) q_{ij} - \frac{1}{4} G^{\mu \nu}_a G^a_{\mu \nu} + \frac{1}{2} \eta^*_a(\delta_{ab} {\Box}^2 + gf_{abc} G^\mu_c \partial_\mu)\eta_a + \frac{1}{2\alpha} (\partial^\mu G_\mu)(\partial^\nu G_\nu),
\end{equation}
em que $D_{\mu} = \partial_\mu + igG_\mu$ é a derivada covariante, $q$ são campos de quarks, $G_\mu$ são campos de gluons e $\eta_a$ são campos fantasmas. 

Para nossas deduções, vamos trabalhar com a métrica euclidiana via rotação de Wick $t \rightarrow -i\tau$:
\begin{align*}
    x_0 &= -ix_4 \\
    \partial_0 &= i\partial_4 \\
    \gamma_0 &= \gamma_4 \\
    \gamma_i &= -i\gamma_i
\end{align*}

e transformar a lagrangiana para o espaço do momento via transformada de Fourier, ignorando os índices $i,j$. Com essas duas mudanças obtemos (ver _momento.ipynb_):
\begin{align}
    \mathcal{L}_{QCD} &= \bar{q} S^{-1}_0(p)q + ig\bar{q}\gamma^\mu G_{a\mu} q + \frac{1}{2} G_{a\mu} (D^{\mu\nu}_0)^{-1} G_{a\nu} - \frac{1}{2} \eta^*_a p^2 \eta_a \nonumber \\
    &+ \frac{1}{4} g^2 (f_{abc}G_{b\mu}G_{c\nu})(f_{ade}G_{d\mu}G_{e\nu}) - \frac{i}{2}g(f_{abc}G_{b\mu}G_{c\nu})(p_\mu G_{a\nu} - p_\nu G_{a\mu}) + \frac{i}{2}gf_{abc}G_{c\mu} \eta^*_a p_\mu \eta_b
\end{align}

A ação da QCD é, portanto,
\begin{equation}
    \mathcal S[q, \bar{q}, G] = \int \frac{\mathrm{d}^4p}{(2\pi)^4} \ \mathcal{L}_{QCD},
\end{equation}
e o funcional gerador da teoria quântica de campos é
\begin{equation}
    Z[\bar{K}, K, J] = \int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J^\mu G_\mu},
\end{equation}
em que algo do tipo $J\phi$ denota o produto interno $J(p) \cdot \phi(p) = \int \frac{\mathrm{d}^4p}{(2\pi)^4} \ J(p) \phi(p)$.

### Derivação das DSEs

Temos como ponto de partida a relação
\begin{equation}
    \int^{+\infty}_{-\infty} dx \ \frac{\partial \rho}{\partial x} = 0 \ \Leftrightarrow \ \rho (\pm \infty) = 0,
    \tag{$\star$}
\end{equation}
que é simplesmente uma aplicação do teorema fundamental do cálculo.

Especificamente, vamos utilizar as três identidades que são casos particulares de $(\star)$:

\begin{align}
    &\int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ \frac{\delta}{\delta q} \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J_\mu G_\mu} = 0 \tag{$*$},\\

    &\int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ \frac{\delta}{\delta G_\nu} \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J_\mu G_\mu} = 0 \tag{$**$},\\

    &\int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ \frac{\delta}{\delta \eta^*} \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J_\mu G_\mu} = 0 \tag{$***$},
\end{align}
em que $q,G_\mu,\eta \rightarrow 0$ no infinito.

Também será útil a identidade funcional
\begin{equation}
    \int \mathcal{D} \phi \ \phi(x) \ e^{-\mathcal{S}[\phi] + J(x) \cdot \phi(x)} = \frac{\delta }{\delta J(x)} \int \mathcal{D} \phi \ e^{-\mathcal{S}[\phi] + J(x) \cdot \phi(x)}.
    \tag{$\star \star$}
\end{equation}

### DSE para o Propagador de Quarks

Começamos com a identidade $(*)$:
\begin{equation*}
    \int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ \frac{\delta}{\delta q} \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J_\mu G_\mu} = 0.
\end{equation*}

Avaliando o lado esquerdo da igualdade (atuando a derivada funcional), temos
\begin{equation*}
    \int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ \left( - \frac{\delta \mathcal{S}}{\delta q} + \bar{K} \right) \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J_\mu G_\mu} = 0
\end{equation*}
que podemos mover para fora da integral pela identidade $(\star \star)$  como
\begin{equation*}
    \left[- \frac{\delta \mathcal{S}}{\delta q} \left(\frac{\delta}{\delta \bar{K}}, \frac{\delta}{\delta K}, \frac{\delta}{\delta J} \right) + \bar{K} \right] \underbrace{\int \mathcal{D}(q, \bar{q}, G, \eta, \eta^*) \ e^{-\mathcal{S} + \bar{q}K + \bar{K}q + J_\mu G_\mu}}_{Z[\bar{K}, K, J]} = 0.
\end{equation*}

Portanto, temos a equação de Dyson-Schwinger, em forma reduzida, no espaço do momento:
\begin{equation}
    \left[ \frac{\delta \mathcal{S}}{\delta q(p)} \left(\frac{\delta}{\delta \bar{K}(p)}, \frac{\delta}{\delta K(p)}, \frac{\delta}{\delta J(p)} \right) - \bar{K}(p) \right] Z[\bar{K}, K, J] = 0.
\end{equation}

\end{equation*}


Podemos substituir a ação, dada pela lagrangiana $\mathcal{L}_{QCD}$, cuja derivada é:

\begin{align*}
    \frac{\delta \mathcal{S}}{\delta q(p)} &= \frac{\delta}{\delta q(p)} \int \frac{\mathrm{d}^4k}{(2\pi)^4} \left\{ \bar{q} S^{-1}_0(k)q + ig\bar{q}\gamma^\mu G_{a\mu} q + \frac{1}{2} G_{a\mu} (D^{\mu\nu}_0)^{-1} G_{a\nu} - \frac{1}{2} \eta^*_a p^2 \eta_a + ...\right\} \\

    &= \int \frac{\mathrm{d}^4k}{(2\pi)^4} \ \left\{ \bar{q} S^{-1}_0(k) (2\pi)^4\delta^{(4)}(p - k) + ig\bar{q}\gamma^\mu G_{a\mu}(k) (2\pi)^4\delta^{(4)}(p - k) \right\} \\

    &= \ \bar{q} S^{-1}_0(p) + ig\bar{q}\gamma^\mu G_{a\mu}(p)
\end{align*}
portanto
\begin{equation*}
    \left[ \frac{\delta \mathcal{S}}{\delta q} \left(\frac{\delta}{\delta \bar{K}}, \frac{\delta}{\delta K}, \frac{\delta}{\delta J} \right) - \bar{K} \right] Z = \left[ S^{-1}_0 \frac{\delta}{\delta K} + ig \gamma^\mu \frac{\delta}{\delta K} \frac{\delta}{\delta J^\mu} - \bar{K} \right] Z = 0.
\end{equation*}

Assim, ficamos com
\begin{equation}
    \left[ S^{-1}_0(p) \frac{\delta}{\delta K(p)} + ig \gamma^\mu \frac{\delta}{\delta K(p)} \frac{\delta}{\delta J^\mu (p)} - \bar{K}(p) \right] Z[\bar{K}, K, J] = 0.
\end{equation}

\begin{equation*}

\end{equation*}

Agora podemos reorganizar e fazer uma derivada adicional em relação a $\bar{K}(q)$:

\begin{align*}
    \frac{\delta}{\delta \bar{K}(q)} \left[ S^{-1}_0(p) \frac{\delta}{\delta K(p)} + ig \gamma^\mu \frac{\delta}{\delta K(p)} \frac{\delta}{\delta J^\mu(p)} \right] Z &= \frac{\delta}{\delta \bar{K}(q)} \bar{K}(p)Z \\ \therefore

    \left[ S^{-1}_0(p) \frac{\delta^2}{\delta K(p) \delta \bar{K}(q)} + ig \gamma^\mu \frac{\delta}{\delta J^\mu(p)} \frac{\delta^2}{\delta K(p) \delta \bar{K}(q)} \right] Z &= \delta^4(q - p)Z + \bar{K}(p) \frac{\delta Z}{\delta \bar{K}(q)}.
\end{align*}

Identificamos aqui a função de dois pontos dos quarks:
\begin{equation}
    S(p, q) = Z^{-1} \frac{\delta^2 Z}{\delta K(p) \delta \bar{K}(q)}.
\end{equation}

Substituindo, obtemos
\begin{equation*}
    \left[ S^{-1}_0(p) + ig \gamma^\mu \frac{\delta}{\delta J^\mu(p)} \right] Z S(p, q) = \delta^4(q - p)Z + \bar{K}(p) \frac{\delta Z}{\delta \bar{K}(q)}
\end{equation*}
que podemos expandir usando a regra do produto:
\begin{equation*}
    \left[ Z S^{-1}_0(p) + ig \gamma^\mu \frac{\delta Z}{\delta J^\mu(p)} + ig Z \gamma^\mu \frac{\delta}{\delta J^\mu(p)} \right] S(p, q) = \delta^4(q - p)Z + \bar{K}(p) \frac{\delta Z}{\delta \bar{K}(q)}.
\end{equation*}

Dividindo por Z:
\begin{equation}
    \left[ S^{-1}_0(p) + ig Z^{-1} \gamma^\mu \frac{\delta Z}{\delta J^\mu(p)} + ig \gamma^\mu \frac{\delta}{\delta J^\mu(p)} \right] S(p, q) = \delta^4(q - p) + \bar{K}(p) Z^{-1} \frac{\delta Z}{\delta \bar{K}(q)}
\end{equation}
em que identificamos o segundo termo do lado esquerdo da equação como o valor esperado do campo de gluons, que deve ser nulo quando a fonte $J^\mu$ vai para zero.

Fazendo $J^\mu, \bar{K} \rightarrow 0$, obtemos
\begin{equation}
    \left[ S^{-1}_0(p) + ig \gamma^\mu \frac{\delta}{\delta J^\mu(p)} \right] S(p, q) = \delta^4(q - p),
\end{equation}
que é a equação de Dyson-Schwinger para o propagador de quarks.

### Vértice de interação Quark-Glúon

Podemos expandir o segundo termo da equação para o propagador de quarks, usando a regra da cadeia:
\begin{equation}
    \frac{\delta S(p, q)}{\delta J_\mu (p)} = \int_k \frac{\delta \langle G_\nu(k) \rangle}{\delta J_\mu (p)} \frac{\delta S(p, q)}{\delta \langle G_\nu(k) \rangle}
\end{equation}
em que representamos $\int_k = \int d^4k / (2\pi)^4$. Podemos ainda definir a ação efetiva
\begin{equation}
    \Gamma[Q, \bar{Q}, \tilde{G}] := W[\bar{K}, K, J] - \int_p (\bar{Q}(p)K(p) + \bar{K}(p)Q(p) + \tilde{G}(p)J(p))
\end{equation}
como a transformada de Legendre do funcional gerador das funções de Green conexas. Com isso, temos que
\begin{equation}
    \frac{\delta^2 W}{\delta \bar{K} (p) \delta K (q)} = \left( \frac{\delta^2 \Gamma}{\delta Q (p) \delta \bar{Q} (q)} \right)^{-1} = \Gamma^{-1}_2 (p, q),
\end{equation}
e o segundo termo então fica
\begin{equation}
    \frac{\delta S(p, q)}{\delta J_\mu (p)} = - \int_k \frac{\delta \tilde{G}_\nu(k)}{\delta J_\mu (p)} \frac{\delta \Gamma^{-1}_2 (p, q)}{\delta \tilde{G}_\nu(k)}.
\end{equation}

Podemos calcular a derivada de $\Gamma^{-1}_2$ da seguinte forma:
\begin{equation*}
    \frac{\delta (\Gamma_2 \Gamma^{-1}_2)}{\delta \tilde{G}_\nu(k)} = 0 \ \ \therefore \ \ \frac{\delta \Gamma_2}{\delta \tilde{G}_\nu(k)} \Gamma^{-1}_2 + \Gamma_2 \frac{\delta \Gamma^{-1}_2}{\delta \tilde{G}_\nu(k)} = 0,
\end{equation*}
então multiplicamos à esquerda por $\Gamma_2^{-1}$ (por que é uma matriz) e reorganizamos:
\begin{equation*}
    \frac{\delta \Gamma^{-1}_2}{\delta \tilde{G}_\nu(k)} = - \Gamma_2^{-1} \frac{\delta \Gamma_2}{\delta \tilde{G}_\nu(k)} \Gamma_2^{-1},
\end{equation*}
portanto, explicitando os momentos,
\begin{equation}
    \frac{\delta \Gamma^{-1}_2 (p, q)}{\delta \tilde{G}(k)} = - \int_{k_1, k_2} S(p, k_1) \frac{\delta \Gamma_2 (k_1, k_2)}{\delta \tilde{G}_\nu(k)} S(k_2, q),
\end{equation}
então
\begin{equation}
    \frac{\delta S(p, q)}{\delta J_\mu (p)} = \int_{k, k_1, k_2} S(k_1) D_{\mu \nu}(k, p) \frac{\delta \Gamma_2 (k_1, k_2)}{\delta \tilde{G}_\nu(k)} S(k_2, q).
\end{equation}
em que identificamos o propagador completo do glúon:
\begin{equation}
    D_{\mu \nu} (k, p) = \frac{\delta \tilde{G}_\nu (k)}{\delta J_\mu (p)}.
\end{equation}



Agora definimos a função de vértice como
\begin{equation}
    ig \Lambda_\nu (p, q; k) := \frac{\delta \Gamma_2 (p, q)}{\delta \tilde{G}_\nu(k)} = \frac{\delta}{\delta \tilde{G}_\nu(k)} \frac{\delta^2 \Gamma}{\delta Q(p) \delta \bar{Q}(q)},
\end{equation}
que nos dá
\begin{equation}
    ig\frac{\delta}{\delta J^\mu(p)} S(p, q) = -g^2 \int_{k, k_1, k_2} S(k_1) D_{\mu \nu}(k, p) \Lambda(k_1,k_2;k) S(k_2, q),
\end{equation}
bem como a função de auto-energia do quark:
\begin{equation}
    \Sigma(p) := g^2 \int_{k_1, k_2} \gamma_\mu S(k_1) D_{\mu \nu}(k_1, k_2) \Lambda(k_1,q;k_2),
\end{equation}
que nos permite reescrever a DSE para o propagador do quark como
\begin{equation}
    S^{-1}_0(p)S(p, q) - \int_k \Sigma(p, k) S(k, q) = (2\pi)^4 \delta^{(4)}(p - q).
\end{equation}

Considerando que
\begin{equation}
    S(p,q) = (2\pi)^4 \delta^{(4)}(p-q)S(p),
\end{equation}
podemos multiplicar ambos os lados por $S_0(p)$ reescrevê-la como
\begin{align*}
    (2\pi)^4 \delta(p-q)S(p) &= S_0(p) \int_k \Sigma(p, k) S(k, q) + S_0(p)(2\pi)^4 \delta^{(4)}(p - q) \\
    &= S_0(p) \int_k (2\pi)^4 \delta^{(4)}(p - k) \Sigma(p) (2\pi)^4 \delta^{(4)}(k - q) S(k) + S_0(p)(2\pi)^4 \delta^{(4)}(p - q) \\
    &= S_0(p) (2\pi)^4 \delta^{(4)}(p - q) \Sigma(p) S(p) + S_0(p)(2\pi)^4 \delta^{(4)}(p - q),
\end{align*}
então, "dividindo" todos os termos por $(2\pi)^4 \delta^{(4)}(p-q)$, a DSE fica
\begin{equation}
    S(p) = S_0(p) + S_0(p)\Sigma(p)S(p).
\end{equation}

Podemos manipular a última expressão:
\begin{align*}
    S(p) &= S_0(p) + S_0(p)\Sigma(p)S(p) \\
    S_0(p) &= S(p) - S_0(p)\Sigma(p)S(p) \\
    1 &= S_0^{-1}(p)S(p) - \Sigma(p)S(p),
\end{align*}
então obtemos a _equação de gap do quark_:
\begin{equation}
    S^{-1}(p) = S_0^{-1}(p) - \Sigma(p).
\end{equation}

\end{equation*}

A equação para a auto-energia, com índices de cor explícitos, fica
\begin{equation}
    \Sigma(p) \mathbb{I} = g^2 \int_k (\gamma_\mu T^a) (S(k)\mathbb{I}) (D_{\mu \nu}(p - k)\delta_{ab}\mathbb{I}) (\Lambda_\nu (k,p;p-k) T^b).
\end{equation}
Usando $T_a T^a = \mathbb{I}(N_c^2 -1)/2N_c$, podemos fatorar a identidade e obter
\begin{equation}
    \Sigma(p) = g^2 \left( \frac{N_c^2 -1}{2N_c} \right) \int_k \gamma_\mu S(k) D_{\mu \nu}(p - k) \Lambda_\nu (k,p;p-k).
\end{equation}

### Aproximação Rainbow

Façamos a função de vértice igual ao vértice nu:
\begin{equation}
    \Lambda_\nu \longrightarrow \gamma_\nu,
\end{equation}
assim, a DSE fica
\begin{equation}
    S^{-1}(p) - S_0^{-1}(p) = -\bar{g}^2 \int_k \gamma_\mu S(k) D^{\mu \nu}(p - k) \gamma_\nu
\end{equation}
em que contraímos $\bar{g}^2 = g^2 \left( \frac{N_c^2 -1}{2N_c} \right)$.

Podemos inserir aqui as formas gerais para o propagador do Glúon:
\begin{align}
    D^{\mu\nu}(k) = \left[ P^{\mu\nu}_T (k) + \alpha P^{\mu\nu}_L (k) \right] \frac{1}{k^2} &= \left[ \Delta_T (k) \left(\delta^{\mu\nu} - \frac{k^\mu k^\nu}{k^2} \right) + \alpha\frac{k^\mu k^\nu}{k^2}  \right] \frac{1}{k^2} \\
    &= \left[ \Delta_T \delta^{\mu\nu} + \left(\alpha - \Delta_T\right) \frac{k^\mu k^\nu}{k^2}  \right] \frac{1}{k^2}
\end{align}
e o propagador do quark:
\begin{equation}
    S(p) = \frac{1}{i p\!\!\!/ A(p^2) + B(p^2)}, \quad S_0(p) = \frac{1}{i p\!\!\!/ + m_0},
\end{equation}
assim,
\begin{equation}
    i p\!\!\!/ A(p^2) + B(p^2) - ip\!\!\!/ - m_0 = -\bar{g}^2 \int_k \gamma_\mu \frac{1}{i k\!\!\!/ A(k^2) + B(k^2)} \left[ \Delta_T \delta^{\mu\nu} + \left(\alpha - \Delta_T\right) \frac{q^\mu q^\nu}{q^2}  \right] \frac{1}{q^2} \gamma_\nu,
\end{equation}
em que $q = p - k$.

Reescrevendo o termo dentro da integral:
\begin{equation}
    \frac{1}{i k\!\!\!/ A(k^2) + B(k^2)} \times \left( \frac{-i k\!\!\!/ A(k^2) + B(k^2)}{-i k\!\!\!/ A(k^2) + B(k^2)} \right) = \frac{-i k\!\!\!/ A(k^2) + B(k^2)}{k^2 A^2(k^2) + B^2(k^2)}
\end{equation}
em que simplesmente multipliquei e dividi pelo complexo conjugado. A DSE então fica
\begin{equation}
    i p\!\!\!/ A(p^2) + B(p^2) - ip\!\!\!/ - m_0 = \bar{g}^2 \int_k \gamma_\mu \frac{i k\!\!\!/ A(k^2) - B(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \left[ \Delta_T \delta^{\mu\nu} + \left(\alpha - \Delta_T\right) \frac{q^\mu q^\nu}{q^2}  \right] \frac{1}{q^2} \gamma_\nu.
\end{equation}

Podemos calcular o traço e obter projeções para $A(p^2)$ e $B(p^2)$:
\begin{align}
    A(p^2) &= 1 - \frac{\bar{g}^2}{p^2} \int_k \frac{A(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \frac{1}{q^2} \left\{\left[(p \cdot k) - 2\frac{(p \cdot q)(k \cdot q)}{q^2} \right]\alpha + \left[(p \cdot k) + 2\frac{(p \cdot q)(k \cdot q)}{q^2} \right]\Delta_T \right\}
    \\
    B(p^2) &= m_0 -\bar{g}^2 \int_k \frac{B(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \frac{1}{q^2} \left( \alpha + 3\Delta_T \right)
\end{align}
(para o cálculo em detalhes, ver o arquivo _traço.ipynb_).

No calibre de Landau, $\alpha = 0$, assim,
\begin{align}
    A(p^2) &= 1 - \frac{\bar{g}^2}{p^2} \int_k \frac{A(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \frac{\Delta_T(q)}{q^2} \left[(p \cdot k) + 2\frac{(p \cdot q)(k \cdot q)}{q^2} \right]
    \\
    B(p^2) &= m_0 -3\bar{g}^2 \int_k \frac{B(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \frac{\Delta_T (q)}{q^2}.
\end{align}

Podemos mudar as variáveis de integração para coordenadas esféricas quadridimensionais, escolhendo $p$ ao longo do eixo $k_1$:
\begin{align*}
    k_1 &= k \cos{\beta}\\
    k_2 &= k \sin{\beta} \cos{\theta}\\
    k_3 &= k \sin{\beta} \sin{\theta} \cos{\phi}\\
    k_4 &= k \sin{\beta} \sin{\theta} \sin{\phi},
\end{align*}
assim
\begin{equation}
    \int_k f(p,k) := \int \frac{d^4k}{(2\pi)^4} f(p,k) \longrightarrow \frac{1}{(2\pi)^4} \int_0^L k^3dk \int_0^\pi \sin^2{\beta} d\beta \int_0^\pi \sin{\theta} d\theta \int_0^{2\pi} d\phi \ f(p,k)
\end{equation}
e já podemos calcular as integrais angulares triviais, deixando
\begin{align}
    \int_k f(p,k) &= \frac{4\pi}{(2\pi)^4} \int_0^L k^3dk \int_0^\pi \sin^2{\beta} d\beta \ f(p^2,k^2, pk\cos{\beta}) \nonumber
    \\
    &= \frac{4\pi}{(2\pi)^4} \int_0^{L^2} \frac{1}{2}tdt \int_0^\pi \sin^2{\beta} d\beta \ f(p^2,k^2, pk\cos{\beta}) \qquad (t=k^2 \therefore dk = 1/2k \ dt) \nonumber
    \\
    &= \frac{1}{(2\pi)^3} \int_0^L k^2dk^2 \int_0^\pi \sin^2{\beta} d\beta \ f(p^2,k^2, pk\cos{\beta}) \qquad (t = k^2 \therefore dt = d(k^2), \ L^2 \rightarrow L)
\end{align}
em que $f$ é uma função escalar, portanto só depende das quantidades invariantes $p^2,k^2,p\cdot k = pk\cos{\beta}$. O limite de integração $L^2$ foi renomeado, não meramente substituido, já que $k^2$ vai de $0$ a $L^2$ (isso é um abuso de notação por parte do Michael).

As equações para $A,B$ ficam
\begin{align}
    A(p^2) &= 1 - \frac{\bar{g}^2}{8\pi^3 p^2} \int_0^L k^2dk^2 \left( \frac{A(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \right) \int_0^\pi \sin^2{\beta} d\beta \left(\frac{\Delta_T(p^2 + k^2 - 2pk\cos{\beta})}{p^2 + k^2 - 2pk\cos{\beta}}\right)
    \nonumber \\
    &\quad \times \left[pk\cos{\beta} + 2\frac{(p^2 + k^2)(pk\cos{\beta}) - p^2k^2 - (pk\cos{\beta})^2}{p^2 + k^2 - 2pk\cos{\beta}} \right]
    \\
    B(p^2) &= m_0 - \frac{3\bar{g}^2}{8\pi^3} \int_0^L k^2dk^2 \left( \frac{B(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \right) \int_0^\pi \sin^2{\beta} d\beta \left(\frac{\Delta_T(p^2 + k^2 - 2pk\cos{\beta})}{p^2 + k^2 - 2pk\cos{\beta}}\right),
\end{align}
sendo que
\begin{align*}
    p \cdot q &= p \cdot (p-k) = p^2 - p \cdot k = p^2 - pk\cos{\beta} \\
    k \cdot q &= k \cdot (p-k) = k \cdot p - k^2 = pk\cos{\beta} - k^2,
\end{align*}
então
\begin{align*}
    \left[(p \cdot k) + 2\frac{(p \cdot q)(k \cdot q)}{q^2} \right] &= \left[pk\cos{\beta} + 2\frac{(p^2 - pk\cos{\beta})(pk\cos{\beta} - k^2)}{p^2 + k^2 - 2pk\cos{\beta}} \right]
    \\
    &= \left[pk\cos{\beta} + 2\frac{(p^2 + k^2)(pk\cos{\beta}) - p^2k^2 - (pk\cos{\beta})^2}{p^2 + k^2 - 2pk\cos{\beta}} \right].
\end{align*}

### Aproximação Quenched
Agora vamos especificar a forma do propagador do Glúon. A aproximação quenched consiste em escolher seu propagador nu, fazendo:
\begin{equation}
    \Delta_T(q) \longrightarrow 1,
\end{equation}
o que nos leva às equações
\begin{align}
    A(p^2) &= 1 - \frac{\bar{g}^2}{8\pi^3 p^2} \int_0^L k^2dk^2 \left( \frac{A(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \right) \int_0^\pi \sin^2{\beta} d\beta \left(\frac{1}{p^2 + k^2 - 2pk\cos{\beta}}\right)
    \nonumber \\
    &\quad \times \left[pk\cos{\beta} + 2\frac{(p^2 + k^2)(pk\cos{\beta}) - p^2k^2 - (pk\cos{\beta})^2}{p^2 + k^2 - 2pk\cos{\beta}} \right]
    \\
    B(p^2) &= m_0 - \frac{3\bar{g}^2}{8\pi^3} \int_0^L k^2dk^2 \left( \frac{B(k^2)}{k^2 A^2(k^2) + B^2(k^2)} \right) \int_0^\pi \sin^2{\beta} d\beta \left(\frac{1}{p^2 + k^2 - 2pk\cos{\beta}}\right),
\end{align}

Podemos fazer uma troca de variáveis em que $x = p^2$ e $y = k^2$, então
\begin{align}
    A(x) &= 1 - \frac{\bar{g}^2}{8\pi^2 p^2} \int_0^\infty ydy \left( \frac{A(y)}{y A^2(y) + B^2(y)} \right) K_A(x,y)
    \\
    B(x) &= m_0 - \frac{3\bar{g}^2}{8\pi^2} \int_0^\infty ydy \left( \frac{B(y)}{y A^2(y) + B^2(y)} \right) K_B(x,y),
\end{align}
em que os kernels de integração são
\begin{align}
    K_A(x,y) &= \frac{1}{\pi} \int_0^\pi \left(\frac{\sin^2{\beta} d\beta}{p^2 + k^2 - 2pk\cos{\beta}}\right)
    \\ 
    &\qquad \times \left[pk\cos{\beta} + 2\frac{(p^2 + k^2)(pk\cos{\beta}) - p^2k^2 - (pk\cos{\beta})^2}{p^2 + k^2 - 2pk\cos{\beta}} \right]
    \\
    K_B(x,y) &= \frac{1}{\pi} \int_0^\pi \left(\frac{\sin^2{\beta} d\beta}{p^2 + k^2 - 2pk\cos{\beta}}\right).
\end{align}


(OBS: Por algum motivo, na tese do Michael, o sinal das integrais está trocado. Talvez seja algo relacionado à convenção de sinal da equação de gap que ele, aparentemente, não usou de forma consistente. Vou adotar a notação dele para a solução numérica, com integrais positivas.)

### solução das Integrais $K_A$ e $K_B$

#### Integrais úteis:

\begin{align}
I_1^0 (x,y) &= \frac{1}{2 \text{max}(x,y)} \\
I_2^0 (x,y) &= \frac{1}{2 |x-y| \text{max}(x,y)} \\
I_1^1 (x,y) &= \frac{\text{min}(x,y)}{4 \text{max}(x,y)} \\
I_2^2 (x,y) &= \frac{xy - 3\text{min}(x,y)}{8 |x-y| \text{max}(x,y)}
\end{align}

#### Kernel $K_A:$

\begin{equation}
    K_A(x,y) = I_1^1 (x,y) + 2(x+y) I_2^1 (x,y) - 2xyI_2^0 (x,y) - 2I_2^2 (x,y)
\end{equation}

Pode-se demonstrar, usando as integrais acima, que $K_A = 0$ para todo $x,y$, portanto, no contexto da aproximação Rainbow-Quenched, $A(x) = Z_2 = 1$.

#### Kernel $K_B:$

\begin{equation}
    K_B(x,y) = I_1^0 (x,y)
\end{equation}



### Solução no eixo real positivo

Para $x$ e $y$ positivos e reais, o kernel $K_B$ é simplificado:
\begin{equation}
    K_B(x,y) = \frac{1}{2\text{max}(x,y)}
\end{equation}
tal que o termo escalar $B$ fica
\begin{equation}
    B(x) = m_0 + \frac{3 \bar{g}^2}{16\pi^2} \int_{\Lambda_{IR}^2}^{\Lambda_{UV}^2} \frac{y \, \mathrm{d}y}{\text{max}(x,y)} 
    \left( \frac{B(y)}{(Z_2)^2y + B^2(y)} \right)
\end{equation}
em que introduzimos cortes (_cuttofs_) no infravermelho e ultravioleta, e $(Z_2)^2=1$.

Essa equação integral acoplada não linear pode ser resolvida numericamente escolhendo uma função de teste para $B(y)$ e resolvendo iterativamente até a solução convergir. O código para a resolução está no arquivo _eixo\_real.ipynb_.
